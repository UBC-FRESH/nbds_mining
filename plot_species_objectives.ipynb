```python
# Equity Silver Baseline Scenario
import matplotlib.pyplot as plt
import os
import numpy as np

# Data for different objectives
labels = ['Hemlock', 'Douglas-fir', 'Aspen', 'Pine', 'Spruce', 'Bal']
sizes_no_harvest = [11.1, 0.7, 1, 45.4, 5.9, 35.9]
sizes_max_hv = [8.2, 0.5, 1.1, 53.1, 4.2, 32.9]
sizes_min_ha = [7.8, 0.9, 1.1, 41.8, 6.3, 42.1]
sizes_max_st = [12.1, 0.9, 1.1, 21, 11.2, 53.7]
sizes_min_em = [8.3, 0.8, 0.9, 51.2, 4.3, 34.5]

# Grouped bar chart data
objectives = ['Max_hv', 'Min_ha', 'Max_st', 'Min_em', 'No_harvest']
data = [sizes_max_hv, sizes_min_ha, sizes_max_st, sizes_min_em, sizes_no_harvest]
data = np.array(data).T

# Directory to save the pdf file
output_dir = './plots/fig'
os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, 'species_bar_chart_equitysilver.pdf')

# Set spacing multiplier
n_species = len(labels)
n_objectives = len(objectives)
bar_width = 0.15
group_spacing = 1.3  # Adjust for more/less space between groups

# Create spaced x positions
x = np.arange(n_species) * group_spacing

# Create bar chart
fig, ax = plt.subplots(figsize=(11, 6))
for i, objective in enumerate(objectives):
    ax.bar(x + i * bar_width, data[:, i], bar_width, label=objective)

# Add labels, title, and legend
ax.set_xlabel('Species')
ax.set_ylabel('Percentage')
ax.set_xticks(x + bar_width * (n_objectives - 1) / 2)
ax.set_xticklabels(labels, rotation=15)  # Added rotation for readability
ax.legend(title="Objectives")

# Adjust layout and save
plt.tight_layout()
plt.savefig(output_file, format='pdf')
plt.show()

print(f"Figure saved as pdf at: {output_file}")
```
