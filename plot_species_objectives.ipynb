```python
import matplotlib.pyplot as plt
import os
import numpy as np

# Data for different objectives
labels = ['Hem', 'DougFir', 'Aspen', 'Pine', 'Spruce', 'Bal']
sizes_initial = [16.6, 0.9, 1.2, 24.5, 6.6, 50.2]
sizes_max_hv = [8.2, 0.5, 1.1, 53.1, 4.2, 32.9]
sizes_min_ha = [7.8, 0.9, 1.1, 41.8, 6.3, 42.1]
sizes_max_st = [12.1, 0.9, 1.1, 21, 11.2, 53.7]
sizes_min_em = [8.3, 0.8, 0.9, 51.2, 4.3, 34.5]

# Grouped bar chart data
objectives = ['Max_hv', 'Min_ha', 'Max_st', 'Min_em', 'Initial']
data = [sizes_max_hv, sizes_min_ha, sizes_max_st, sizes_min_em, sizes_initial ]
data = np.array(data).T

# Directory to save the SVG file
output_dir = './plots/fig'
os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, 'species_bar_chart.svg')

# Create bar chart
x = np.arange(len(labels))  # the label locations
width = 0.2  # the width of the bars

fig, ax = plt.subplots(figsize=(10, 6))
for i, objective in enumerate(objectives):
    ax.bar(x + i * width, data[:, i], width, label=objective)

# Add labels, title, and legend
ax.set_xlabel('Species')
ax.set_ylabel('Percentage')
# ax.set_title('Percentage of Species Under Different Objectives')
ax.set_xticks(x + width * (len(objectives) - 1) / 2)
ax.set_xticklabels(labels)
ax.legend(title="Objectives")

# Adjust layout and save
plt.tight_layout()
plt.savefig(output_file, format='svg')
plt.show()

print(f"Figure saved as SVG at: {output_file}")
```
