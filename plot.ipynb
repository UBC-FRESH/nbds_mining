```python
#Equity Silver max_hv hwp=1, dis =1


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from math import pi

# Define the data
data = {
    "Scenarios": ["BAU", "No_cons", "AAC_80%", "AAC_50%"],
    "Harvest Volume": [9.5385144e+07, 1.3344616e+08, 7.6308117e+07, 4.7692577e+07],
    "Negative Net Emission": [-11327030, -21482762, -16804527, -25123867],
    "Social Indicator": [100842, 141080, 80673, 50421],
    "Economic Indicator": [3438933, 4811151, 2751146, 1719466],
    "Species Indicator": [0.7329, 0.7509, 0.7514, 0.7375],
    "Old Growth Indicator": [196384, 166327, 247449, 367419]
}

# Load data into a DataFrame
df = pd.DataFrame(data)

# Manually normalize each column (except 'Scenarios')
for column in df.columns[1:]:
    min_val = df[column].min()
    max_val = df[column].max()
    df[column] = (df[column] - min_val) / (max_val - min_val)

# Prepare data for radar chart
categories = list(df.columns[1:])
N = len(categories)

# Radar chart setup
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]  # Close the circle by adding the first angle at the end

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Plot each scenario
for i, row in df.iterrows():
    values = row[1:].tolist()
    values += values[:1]  # Close the loop
    ax.plot(angles, values, label=row['Scenarios'], linewidth=2)
    ax.fill(angles, values, alpha=0.25)

# Customize chart
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=10)
plt.title("Radar Chart of Scenarios", fontsize=15, pad=20)

# Adjust legend position and styling
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1))

# Show the plot
plt.show()

```
