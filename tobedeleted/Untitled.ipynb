```python
import matplotlib.pyplot as plt

colors = ["#FF0000", "#FF8C00", "#FFD700", "#00FF00", "#00FFFF", "#1E90FF", "#9400D3", "#FF00FF", "#FF69B4", "#40E0D0"]
labels = [f"Color {i+1}" for i in range(len(colors))]

plt.figure(figsize=(8, 4))
plt.bar(labels, [1] * len(colors), color=colors)
plt.xticks(rotation=45)
plt.show()
```
