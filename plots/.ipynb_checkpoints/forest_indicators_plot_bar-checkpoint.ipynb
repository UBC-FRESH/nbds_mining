```python
# Bar chats
import os
import pandas as pd
import matplotlib.pyplot as plt

def plot_scenarios(csv_files, case_study, obj_mode, scenario_name):
    """
    Plot data from multiple CSV files into a single figure with shared Y-axis ranges for columns.
    
    Parameters:
        csv_files (list of str): List of CSV file paths to read.
        case_study (str): Case study name for output file naming.
        obj_mode (str): Objective mode for output file naming.
        scenario_name (list of str): List of scenario names corresponding to the CSV files.
    """
    # Initialize figure and axes for subplots
    fig, axes = plt.subplots(len(csv_files), 3, figsize=(12, 4 * len(csv_files)))
    
    # Read data and collect min/max values for shared y-axis
    all_oha, all_ohv, all_ogs = [], [], []
    dfs = [pd.read_csv(file) for file in csv_files]
    
    for df in dfs:
        all_oha.extend(df.oha)
        all_ohv.extend(df.ohv)
        all_ogs.extend(df.ogs)
    
    # Determine common y-axis ranges
    common_ylim_oha = (0, max(all_oha) * 1.1)
    common_ylim_ohv = (0, max(all_ohv) * 1.1)
    common_ylim_ogs = (0, max(all_ogs) * 1.1)
    
    # Plot each DataFrame
    for i, (df, obj_mode) in enumerate(zip(dfs, obj_mode)):
        # Plot harvested area (ha)
        axes[i, 0].bar(df.period, df.oha)
        axes[i, 0].set_ylim(common_ylim_oha)
        axes[i, 0].set_title(f'{obj_mode}: Harvested Area (ha)')
        
        # Plot harvested volume (m3)
        axes[i, 1].bar(df.period, df.ohv)
        axes[i, 1].set_ylim(common_ylim_ohv)
        axes[i, 1].set_title(f'{obj_mode}: Harvested Volume (m3)')
        
        # Plot growing stock (m3)
        axes[i, 2].bar(df.period, df.ogs)
        axes[i, 2].set_ylim(common_ylim_ogs)
        axes[i, 2].set_title(f'{obj_mode}: Growing Stock (m3)')
        
        # Add labels for y-axis
        if i == len(csv_files) - 1:  # Add x-axis labels only to the last row
            axes[i, 0].set_xlabel("Period")
            axes[i, 1].set_xlabel("Period")
            axes[i, 2].set_xlabel("Period")
        axes[i, 0].set_ylabel("Value")
    
    # Adjust layout
    plt.tight_layout()
    
    # Save the figure
    folder_path = os.path.join('./fig', case_study)
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
    file_name = f"{case_study}_{scenario_name}_combined_scheduling.svg"
    file_path = os.path.join(folder_path, file_name)
    plt.savefig(file_path, format='svg')
    plt.show()
    plt.close()
    print(f"Plot saved to {file_path}")
```

```python
csv_files = ["equitysilver_max_hv_no_cons_compile_scenario.csv", "equitysilver_max_st_no_cons_compile_scenario_maxstock.csv", "equitysilver_min_em_no_cons_compile_scenario_minemission.csv", 
             "equitysilver_min_ha_no_cons_compile_scenario.csv"]
obj_mode = ["Max_hv", "Max_st", "Min_em", "Min_ha"]
plot_scenarios(csv_files, "equitysilver", obj_mode, scenario_name='No_cons')
```
