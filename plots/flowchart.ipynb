```python
from graphviz import Digraph

# Create a directed graph
flowchart = Digraph(format='png', engine='dot')
flowchart.attr(rankdir='LR', splines='true')

# General style
flowchart.attr('node', shape='box', style='rounded, filled', color='lightblue')

# Data Preparation section
flowchart.node('SelectArea', 'Select area of interest')
flowchart.node('VRI', 'VRI')
flowchart.node('OtherData', 'Other data (HWP pools, displacement effect, etc.)')
flowchart.node('Preparation', 'Preparation')
flowchart.node('YieldCurves', 'Yield curves')
flowchart.node('ProcessOldness', 'Oldness curve')
flowchart.node('ProcessCarbon', 'Carbon curves')
flowchart.node('libcbm', 'libcbm', shape='cylinder', color='lightgoldenrod1')
flowchart.node('ForestInventory', 'Forest inventory resource', shape='cylinder', color='lightgoldenrod1')
flowchart.node('ModelInput', 'Model input data', shape='cylinder', color='lightgoldenrod1')

# Model Execution section
flowchart.node('ws3', 'ws3')
flowchart.node('ScenarioSelection', 'Scenario selection')
flowchart.node('ObjectiveMode', 'Objective mode selection')
flowchart.node('Optimize', 'Optimize')

# Outputs section
flowchart.node('Outputs', 'Outputs', shape='cylinder', color='lightgoldenrod1')
flowchart.node('Harvesting', 'Harvesting volumes and areas')
flowchart.node('CarbonStocks', 'Carbon stocks and emissions')
flowchart.node('SocioIndicators', 'Socioeconomic indicators')
flowchart.node('Biodiversity', 'Biodiversity indicators')

# Data Preparation connections
flowchart.edges([
    ('SelectArea', 'VRI'),
    ('VRI', 'ForestInventory'),
    ('OtherData', 'Preparation'),
    ('Preparation', 'YieldCurves'),
    ('YieldCurves', 'ProcessOldness'),
    ('YieldCurves', 'ProcessCarbon'),
    ('libcbm', 'ProcessOldness'),
    ('libcbm', 'ProcessCarbon'),
    ('ForestInventory', 'Preparation'),
    ('ProcessOldness', 'ModelInput'),
    ('ProcessCarbon', 'ModelInput'),
])

# Model Execution connections
flowchart.edges([
    ('ModelInput', 'ws3'),
    ('ws3', 'ScenarioSelection'),
    ('ScenarioSelection', 'ObjectiveMode'),
    ('ObjectiveMode', 'Optimize'),
    ('Optimize', 'Outputs'),
])

# Outputs connections
flowchart.edges([
    ('Outputs', 'Harvesting'),
    ('Outputs', 'CarbonStocks'),
    ('Outputs', 'SocioIndicators'),
    ('Outputs', 'Biodiversity'),
])

# Save and render the diagram
flowchart.render('forest_flowchart', view=True)
```
